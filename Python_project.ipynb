```python
#PART2:NEW COLUMN CREATION
```

```python
print(df.columns)
```

```python
#1.Extract year from order date
df['Order_Year'] = pd.to_datetime(df['order_date']).dt.year
print(df)
```

```python
#2.Extract month from order date
df['Order_Month'] = pd.to_datetime(df['order_date']).dt.month
print(df)
```

```python
#3.Extract month name from order date
df['Order_Month_Name'] = pd.to_datetime(df['order_date']).dt.month_name()
print(df)
```

```python
#4.Extract day from order date
df['Order_Day'] = pd.to_datetime(df['order_date']).dt.day
print(df)
```

```python
#5.create a column called Revenue Category:
#a.High revenue(>1000)
#b.Medium revenue(500-1000)
#c.Low revenue(<500)

def revenue_category(revenue):
    if revenue>1000:
      return "High Revenue"
    elif revenue>=500:
      return "Medium Revenue"
    else:
      return "Low Revenue"
df['revenue_category']=df['total_revenue'].apply(revenue_category)
print(df[['total_revenue','revenue_category']].head())
```

```python
#6.Create a column called Customer Satisfaction:
#a.Excellent(Rating>=4.5)
#b.Good(Rating3-4.4)
#c.Poor(<3)
def customer_satisfaction(rating):
    if rating>=4.5:
      return "Excellent"
    elif rating>=3:
      return "Good"
    else:
      return "Poor"
df['customer_satisfaction']=df['rating'].apply(customer_satisfaction)
print(df[['rating','customer_satisfaction']].head())
```
